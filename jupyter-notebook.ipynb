```python
from dotenv import load_dotenv
import anthropic

load_dotenv()
client = anthropic.Anthropic()

respuesta = client.messages.create(
    model="claude-sonnet-4-5",
    max_tokens=100,
    messages=[{"role": "user", "content": "Hola, ¿funciona?"}]
)
print(respuesta.content[0].text)
```
